```python

import numpy as np
import matplotlib.pyplot as plt
```

```python
#Load the data into a numpy array
ecg = np.loadtxt("ECGu.txt")
```

```python
#Look at the data
print("Data loaded correctly.")
```

```python
print(ecg[:10])
```

```python
# we need to plot the raw ECG data, so we will use matplotlib libraryto do that
# we want to make sure to label the axes to the plot
import matplotlib.pyplot as plt
```

```python
#check visually if the plot is correctly labeled
```

```python
#conversion to mV
ecg_mV = ecg * 1024
```

```python
# Conversion to seconds using the sampling frequency
sampling_frequency = 1000  # Hz
timestamps = np.arange(ecg.shape[0]) / sampling_frequency
```

```python
# Plot using seconds on the x-axis
plt.plot(timestamps, ecg_mV)
plt.xlabel("Time (s)")
plt.ylabel("ECG (mV)")
plt.title("ECG Signal")
plt.xticks(np.arange(0, timestamps[-1] + 1, 1))
plt.show()
```

By looking at the graph we can comfirm that the nubers of timeseries and samples are correct.
We do have the 3 "steps" of a heart contraction.1-> Left and right atrium depolarisation; 2-> left/right ventricular depolarisartion; 3-> ventricular re-polarisation
