```python
# we need to plot the raw ECG data, so we will use matplotlib libraryto do that
# we want to make sure to label the axes to the plot
import matplotlib.pyplot as plt
```

```python
#check visually if the plot is correctly labeled
```

```python
#conversion to mV
ecg_mV = ecg * 1024
```

```python
# Conversion to seconds: one sample every millisecond
timestamps = np.arange(len(ecg)) / 1000
```

```python
# Plot using seconds on the x-axis
plt.plot(timestamps, ecg_mV)
plt.xlabel("Time (s)")
plt.ylabel("Voltage (mV)")
plt.title("ECG Signal")
plt.xticks(np.arange(0, timestamps[-1] + 1, 1))
plt.show()
```
